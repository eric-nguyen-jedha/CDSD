```python
import textwrap

# Fonction pour insérer des <br> toutes les 40 lettres
def wrap_text(text, width=40):
    if pd.isna(text):  # Vérifie si le texte est vide
        return ""
    wrapped = textwrap.wrap(text, width=width)
    return "".join(wrapped)  # Utilise <br> pour les sauts de ligne HTML

# Applique cette fonction à la colonne 'hotel_description'
result['hotel_description_hover_1'] = result['hotel_description'].apply(wrap_text, width=40)
```

```python
# Dictionnaire d'emojis
facilities_emoji_map = {
    "Ascenceur" : "🛗",
    "Connexion Wi-Fi gratuite": "📶",
    "Restaurant": "🍽",
    "Parking gratuit" : "🚗",
    "Parking": "🚗",
    "Parking privé": "🚗",
    "Réception ouverte 24h/24" : "🕚",
    "En bord de plage": "🏖",
    "Chambres non-fumeurs":"🚭",
    "Terrasse": "🩴",
    "Jardin" : "🌱",
    "Chambres familiales": "👨‍👩‍👧‍👦",
    "Table à manger": "🍽️",
    "Piscine extérieure" : "🏊",
    "Machine à café": "☕"
}

# --- ÉTAPE 1 : Définir la liste des colonnes d'équipements ---
facility_cols = ['hotel_facilities', 'hotel_facilities_1', 'hotel_facilities_2', 'hotel_facilities_3', 'hotel_facilities_4']

# --- ÉTAPE 2 : Créer la fonction qui traite une SEULE LIGNE ---
def format_facilities_from_row(row, columns_to_check, emoji_map):
    """
    Prend une ligne de DataFrame, vérifie les colonnes d'équipements,
    et retourne une chaîne HTML formatée.
    """
    formatted_list = []
    for col in columns_to_check:
        facility_name = row[col]
        # On vérifie que la valeur n'est pas nulle/NaN avant de continuer
        if pd.notna(facility_name):
            emoji = emoji_map.get(facility_name, '✅') # Emoji par défaut si non trouvé
            formatted_list.append(f"{emoji} {facility_name}")
            
    return "<br>".join(formatted_list)

# --- ÉTAPE 3 : Appliquer la fonction à chaque ligne pour créer la nouvelle colonne ---
result['facilities_html'] = result.apply(
    lambda row: format_facilities_from_row(row, facility_cols, facilities_emoji_map),
    axis=1
)
```

### Code pour envoyer de PG Admin puis à Redshift ensuite

```python
import pandas as pd
import psycopg2
from psycopg2.extensions import ISOLATION_LEVEL_AUTOCOMMIT

# 1. Configuration de la connexion PostgreSQL (à adapter)
db_config = {
    "host": "localhost",
    "port": "5433",  # Port redirigé depuis Minikube
    "user": "postgres",  # Utilisateur par défaut pour Bitnami
    "password": "",  # Remplace par le mot de passe récupéré
}

# 2. Nom de la base de données et du fichier CSV
db_name = "hotels_db"  # Nom de la base à créer
local_file = "towns_temp_forecast_best_hotels.csv"  # Chemin vers ton fichier CSV

# 3. Lire le fichier CSV
try:
    df = pd.read_csv(local_file)
    print("Fichier CSV chargé avec succès.")
except Exception as e:
    print(f"Erreur lors de la lecture du CSV : {e}")
    raise

# 4. Connexion à PostgreSQL pour créer la base de données
try:
    # Connexion initiale pour créer la base de données
    conn = psycopg2.connect(
        host=db_config["host"],
        port=db_config["port"],
        user=db_config["user"],
        password=db_config["password"]
    )
    conn.set_isolation_level(ISOLATION_LEVEL_AUTOCOMMIT)
    cursor = conn.cursor()

    # Créer la base de données si elle n'existe pas
    cursor.execute(f"SELECT 1 FROM pg_database WHERE datname = '{db_name}';")
    exists = cursor.fetchone()
    if not exists:
        cursor.execute(f"CREATE DATABASE {db_name};")
        print(f"Base de données '{db_name}' créée.")
    else:
        print(f"Base de données '{db_name}' existe déjà.")

    cursor.close()
    conn.close()
except Exception as e:
    print(f"Erreur lors de la création de la base de données : {e}")
    raise

# 5. Connexion à la base de données pour créer la table et insérer les données
try:
    # Connexion à la base de données spécifique
    conn = psycopg2.connect(
        host=db_config["host"],
        port=db_config["port"],
        user=db_config["user"],
        password=db_config["password"],
        database=db_name
    )
    cursor = conn.cursor()

    # Créer la table
    create_table_query = """
    CREATE TABLE IF NOT EXISTS hotels_forecast (
        id SERIAL PRIMARY KEY,
        hotel_name VARCHAR(255),
        hotel_lat FLOAT,
        hotel_long FLOAT,
        hotel_rating FLOAT,
        hotel_description TEXT,
        hotel_facilities TEXT,
        hotel_facilities_1 TEXT,
        hotel_facilities_2 TEXT,
        hotel_facilities_3 TEXT,
        hotel_facilities_4 TEXT,
        towns VARCHAR(100),
        town_lat FLOAT,
        town_long FLOAT,
        temperature_forecast FLOAT,
        weather_forecast VARCHAR(100)
    );
    """
    cursor.execute(create_table_query)
    conn.commit()
    print("Table 'hotels_forecast' créée ou déjà existante.")

    # Insérer les données
    for _, row in df.iterrows():
        insert_query = """
        INSERT INTO hotels_forecast (
            hotel_name, hotel_lat, hotel_long, hotel_rating, hotel_description,
            hotel_facilities, hotel_facilities_1, hotel_facilities_2, hotel_facilities_3, hotel_facilities_4,
            towns, town_lat, town_long, temperature_forecast, weather_forecast
        )
        VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s, %s);
        """
        cursor.execute(insert_query, (
            row['hotel_name'], row['hotel_lat'], row['hotel_long'], row['hotel_rating'], row['hotel_description'],
            row['hotel_facilities'], row['hotel_facilities_1'], row['hotel_facilities_2'],
            row['hotel_facilities_3'], row['hotel_facilities_4'], row['towns'], row['town_lat'],
            row['town_long'], row['temperature_forecast'], row['weather_forecast']
        ))
    conn.commit()
    print("Données insérées avec succès dans PostgreSQL.")

except Exception as e:
    print(f"Erreur lors de l'insertion dans PostgreSQL : {e}")
    if 'conn' in locals():
        conn.rollback()
finally:
    if 'conn' in locals():
        cursor.close()
        conn.close()
    print("Connexion fermée.")
```

```python
import psycopg2

# Configuration de la connexion (à adapter)
db_config = {
    "host": "localhost",
    "port": "5433",  # Port redirigé depuis Minikube
    "user": "postgres",  # Utilisateur par défaut pour Bitnami
    "password": "",  # Remplace par ton mot de passe
    "database": "hotels_db"  # Nom de la base de données
}

def tester_connexion_postgresql():
    try:
        # 1. Connexion à la base de données
        conn = psycopg2.connect(**db_config)
        cursor = conn.cursor()
        print("✅ Connexion à PostgreSQL réussie !")

        # 2. Vérifier que la table `hotels_forecast` existe
        cursor.execute("""
            SELECT EXISTS (
                SELECT FROM information_schema.tables
                WHERE table_name = 'hotels_forecast'
            );
        """)
        table_exists = cursor.fetchone()[0]
        if table_exists:
            print("✅ La table 'hotels_forecast' existe.")

            # 3. Compter le nombre de lignes dans la table
            cursor.execute("SELECT COUNT(*) FROM hotels_forecast;")
            row_count = cursor.fetchone()[0]
            print(f"📊 Nombre de lignes dans la table : {row_count}")

            # 4. Afficher les 5 premières lignes (optionnel)
            if row_count > 0:
                cursor.execute("SELECT * FROM hotels_forecast LIMIT 5;")
                colonnes = [desc[0] for desc in cursor.description]
                print("\n🔍 Aperçu des données :")
                print(", ".join(colonnes))
                for row in cursor.fetchall():
                    print(row)
        else:
            print("❌ La table 'hotels_forecast' n'existe pas.")

    except psycopg2.Error as e:
        print(f"❌ Erreur de connexion ou de requête : {e}")
    finally:
        if 'conn' in locals():
            cursor.close()
            conn.close()
            print("🔌 Connexion fermée.")

# Exécuter le test
tester_connexion_postgresql()
```
