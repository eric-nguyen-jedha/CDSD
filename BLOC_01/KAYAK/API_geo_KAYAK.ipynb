### Récupération de coordonées GPS pour une trentaines de villes françaises

```python
# Packages 
import requests
import pandas as pd
import asyncio
from geopy.geocoders import Nominatim
import time
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
# Accéder à la variable appid
appid = os.getenv('appid')
```

```python
town_name = ["Mont Saint Michel","Saint Malo","Bayeux","Le Havre","Rouen","Paris","Amiens","Lille","Strasbourg","Chateau du Haut Koenigsbourg","Colmar","Eguisheim","Besancon","Dijon","Annecy","Grenoble","Lyon","Gorges du Verdon","Bormes les Mimosas","Cassis","Marseille",
"Aix en Provence","Avignon","Uzes","Nimes","Aigues Mortes","Saintes Maries de la mer","Collioure","Carcassonne","Ariege","Toulouse","Montauban","Biarritz","Bayonne","La Rochelle"]

geolocator = Nominatim(user_agent="eric_jedha")
location = geolocator.geocode("Lille")
best_towns = []

for geo in town_name:
  geolocator = Nominatim(user_agent="eric_jedha")
  location = geolocator.geocode(geo)
  time.sleep(5)
  best_towns.append({'Towns' : location[0].split(",")[0], 
                            'Latitude' : location[1][0],
                            'Longitude' : location[1][1],
                 })
```

```python
best_towns 
```

```python
df = pd.DataFrame(best_towns)
```

```python
#Créer un index pour le dataframe
df['Id'] = [x for x in range(0, 35)]
df
```

```python
# Sauvegarde en CSV pour éviter de refaire les requêtes à l'API à chaque fois 
df.to_csv('kayak_town_geo.csv')
```

```python
df_gps = df.copy()
```

```python
# Préparation des listes de coordonnées sans les noms des villes pour l'API Open Weather

towns = df_gps[['Longitude', 'Latitude']]
long_list = towns['Longitude'].values.tolist()
lat_list = towns['Latitude'].values.tolist()
print(long_list[0:35])
print(lat_list[0:35])
```

```python
towns.shape
```

```python
# API Open Weather 



town_and_weather = []

for i in range(35):

    url = f"https://api.openweathermap.org/data/2.5/forecast?lat={lat_list[i]}&lon={long_list[i]}&cnt=7&appid={appid}"
    response = requests.get(url = url)
    response.json()
    icon= ''
    temps = response.json()
    weather7 = temps["list"][0]['weather'][0]['main']
    icon7 = temps["list"][0]['weather'][0]['icon']
    temperature = temps["list"][0]['main']['temp'] 
    temp7 = round(temperature - 273.15)

    town_and_weather.append({'Id' : i,
                            'temperature_forcast' : temp7,
                            'weather_forcast' : weather7})

print(town_and_weather)


```

```python
print(town_and_weather)
```

```python
towns_and_wforcast = pd.DataFrame.from_dict(town_and_weather)
towns_and_wforcast.to_csv('kayak_forecast.csv')
```

```python
df1 = pd.read_csv('kayak_town_geo.csv')
df2 = pd.read_csv('kayak_forecast.csv')
df3 = pd.merge(df1, df2, on = 'Id')

df3 = df3.drop(['Unnamed: 0_x', 'Unnamed: 0_y'], axis = 1)
id_move = df3.pop('Id')
df3.insert(0, 'Id', id_move)

# 10 Les meileures villes à visiter en termes de douceur du climent (les données sont récoltées le 4 février, les prévisions ont été faites pour le 11 février 2025)

best_towns = df3.sort_values(by='temperature_forcast', ascending=False)

best_towns[0:10]
df3.to_csv('Town_Geo_Forcast.csv')
```
