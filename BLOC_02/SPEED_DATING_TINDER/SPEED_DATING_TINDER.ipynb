```python
df['age'] = df['age'].fillna(df['age'].median())
df['field'] = df['field'].fillna('Non spécifié')
```

```python
# Vérifier les colonnes nécessaires
if 'age' in df.columns and 'gender' in df.columns:

    # Créer une version catégorielle du genre pour l'affichage
    df['gender_label'] = df['gender'].map({0: 'FEMALE', 1: 'MALE'})

    # Histogramme interactif avec Plotly Express
    fig = px.histogram(df,
                       x='age',
                       color='gender_label',
                       nbins=20,
                       title="Distribution de l'âge selon le sexe",
                       labels={'age': 'Âge', 'gender_label': 'Sexe'},
                       category_orders={'gender_label': ['Femme', 'Homme']},
                       opacity=0.8,
                       color_discrete_map={'FEMALE': FEMALE, 'MALE': MALE},
                       width=SWIDTH,  # Largeur en pixels
                        height=SHEIGHT, # Hauteur en pixels,  # Spécifier les couleurs ici
)
    fig.update_layout(barmode='overlay')
    fig.show()
else:
    print("Les colonnes 'age' et/ou 'gender' sont manquantes dans le dataset.")
```

```python

# Vérifier que la colonne 'race' existe
if 'race' in df.columns:

    # Compter la fréquence de chaque race
    race_counts = df['race'].value_counts().reset_index()
    race_counts.columns = ['Race', 'Count']

    # Mapping correct des races
    race_labels = {
        1: 'Black',
        2: 'European',
        3: 'Latino',
        4: 'Asian',
        5: 'Native American',
        6: 'Other'
    }

    # Remplacer les codes par les labels
    race_counts['Race'] = race_counts['Race'].map(race_labels).fillna('Inconnu')

    # Calculer les pourcentages
    race_counts['Pourcentage'] = (race_counts['Count'] / race_counts['Count'].sum()) * 100
    race_counts['Pourcentage'] = race_counts['Pourcentage'].round(2)

    # Afficher la répartition
    print("Répartition des races (en pourcentage) :")
    print(race_counts[['Race', 'Count', 'Pourcentage']])

    # Camembert interactif avec Plotly
    fig = px.pie(
        race_counts,
        names='Race',
        values='Count',
        title="Répartition des ethnies parmi les participants",
        color_discrete_sequence=px.colors.qualitative.Plotly,
       width=SWIDTH,  # Largeur en pixels
        height=SHEIGHT,
    )

    fig.update_traces(textinfo='percent+label')
    fig.show()

else:
    print("La colonne 'race' n’existe pas dans le dataset. Vérifie le nom des colonnes.")
```

```python
# Création d'un score global d'attractivité
trait_cols = ['attr1_1', 'sinc1_1', 'intel1_1', 'fun1_1', 'amb1_1']
df['total_attraction_score'] = df[trait_cols].sum(axis=1)
#Taux de match par participant
df['match_ratio'] = df.groupby('iid')['match'].transform(lambda x: x.mean())
```

```python
#Before Dating
# mettre .copy() pour éviter l'alerte d'erreur
before_dating = df[['gender', "attr1_1", "sinc1_1", "intel1_1", "fun1_1", "amb1_1", "shar1_1"]].copy()

# renommer les colonnes pour obtenir une courbe avec des légendes adaptées
before_dating.rename(columns={"gender": "Gender", "attr1_1" : "Attractivité", "sinc1_1" : "Sincérité", "intel1_1" : "Intelligence", "fun1_1" : "Fun", "amb1_1" : "Ambition", "shar1_1": "Centres d'intérêt communs"}, inplace=True)

# Calculer la moyenne par genre
df_mean = before_dating.groupby('Gender').mean().reset_index()

# Pivoter la table
df_melted = df_mean.melt(id_vars=['Gender'], var_name='Trait', value_name='Moyenne')
df_melted = df_melted.sort_values(by="Moyenne", ascending=False)

# Créer une liste des critères triés
sorted_traits = df_melted['Trait'].unique()

# Création du graphique avec `barmode='group'`
fig = px.bar(df_melted, x='Trait', y='Moyenne', color='Gender', barmode='group',
             title="Les critères souhaités avant le Dating", opacity=0.9, 
             labels={'Moyenne': 'Mean Of Points', 'Trait': 'Les critères notées'},
             category_orders={"Trait": sorted_traits}, width=1400, height=600, color_discrete_map={'FEMALE': FEMALE, 'MALE': MALE})

# Forcer un espace entre les barres pour éviter l'empilement visuel
fig.update_layout(bargap=0.2)

# Affichage
fig.show()
```

```python
# Mapping des races selon le .doc
race_labels = {
    1: 'Black',
    2: 'European',
    3: 'Latino',
    4: 'Asian',
    5: 'Native American',
    6: 'Other'
}

# Filtrer les matchs mutuels où les deux personnes ne sont pas de la même race
df_match = df[df['match'] == 1]  # Match mutuel
df_cross_race = df_match[df_match['samerace'] == 0]  # Races différentes

# Remplacer les codes par des labels
df_cross_race['race_label'] = df_cross_race['race'].map(race_labels)
df_cross_race['race_o_label'] = df_cross_race['race_o'].map(race_labels)

# Supprimer les lignes avec race ou race_o manquantes (NaN)
df_cross_race.dropna(subset=['race_label', 'race_o_label'], inplace=True)

# Créer une combinaison raciale (participant x partenaire) - triée pour éviter les doublons comme "A x B" et "B x A"
df_cross_race['Combination'] = df_cross_race[['race_label', 'race_o_label']].apply(lambda x: ' x '.join(sorted(x)), axis=1)

# Créer un DataFrame groupé par genre et combinaison raciale
combination_df = df_cross_race.groupby(['gender', 'Combination']).size().reset_index(name='Count')

# Calculer le total par genre pour normaliser les pourcentages
total_by_gender = combination_df.groupby('gender')['Count'].transform('sum')
combination_df['Pourcentage'] = (combination_df['Count'] / total_by_gender) * 100
combination_df['Pourcentage'] = combination_df['Pourcentage'].round(2)

# Camembert (sunburst) interactif par genre et combinaison raciale
fig_sunburst = px.sunburst(
    combination_df,
    path=['gender', 'Combination'],
    values='Count',
    title="Répartition des combinaisons des ethnies dans les matchs de ethnies différentes (ventilé par genre)",
    color='gender',
    color_discrete_map={'FEMALE': '#FF69B4', 'MALE': '#1E90FF'}
)

fig_sunburst.show()

# Graphique à barres groupées
fig_bar = px.bar(
    combination_df,
    x='Combination',
    y='Pourcentage',
    color='gender',
    barmode='group',
    title="Matchs de ethnies différentes – Répartition en % par genre",
    labels={'Pourcentage': 'Pourcentage (%)', 'Combination': 'Combinaison raciale'},
    color_discrete_map={'FEMALE': '#FF69B4', 'MALE': '#1E90FF'},
    width=1000,
    height=600
)

fig_bar.update_layout(xaxis_tickangle=-45, legend_title_text='Genre')
fig_bar.show()
```

```python
import pandas as pd
import plotly.express as px



# Colonnes d'auto-perception (ce que je pense de moi)
self_cols = {
    'attr3_1': 'Attractivité',
    'sinc3_1': 'Sincérité',
    'intel3_1': 'Intelligence',
    'fun3_1': 'Fun',
    'amb3_1': 'Ambition'
}

# Colonnes de perception extérieure (ce que les autres pensent de moi)
others_cols = {
    'attr5_1': 'Attractivité',
    'sinc5_1': 'Sincérité',
    'intel5_1': 'Intelligence',
    'fun5_1': 'Fun',
    'amb5_1': 'Ambition'
}

# Créer les deux DataFrames
self_ratings = df[list(self_cols.keys())].copy()
self_ratings.rename(columns=self_cols, inplace=True)
self_ratings = self_ratings.mean().reset_index()
self_ratings.columns = ['Critère', 'Moyenne_Moi']

others_ratings = df[list(others_cols.keys())].copy()
others_ratings.rename(columns=others_cols, inplace=True)
others_ratings = others_ratings.mean().reset_index()
others_ratings.columns = ['Critère', 'Moyenne_Autres']

# Fusionner les deux DataFrames
comparison_df = pd.merge(self_ratings, others_ratings, on='Critère')

# Calculer la différence en valeur absolue et en pourcentage
comparison_df['Écart_absolu'] = abs(comparison_df['Moyenne_Moi'] - comparison_df['Moyenne_Autres'])
comparison_df['Écart_en_pourcentage'] = (comparison_df['Moyenne_Moi'] / comparison_df['Moyenne_Autres'] - 1) * 100
# Si "Moyenne_Autres" = 0, éviter la division par zéro
comparison_df['Écart_en_pourcentage'] = comparison_df['Écart_en_pourcentage'].replace([float('inf'), float('-inf')], 0)

# Afficher le résultat
print("Comparaison entre ce que l'on pense de soi et ce que les autres pensent :")
print(comparison_df.round(2))

# Afficher un graphique à barres groupées
# Afficher un graphique à barres groupées
fig = px.bar(
    comparison_df.melt(id_vars='Critère', value_vars=['Moyenne_Moi', 'Moyenne_Autres']),
    x='Critère',
    y='value',
    color='variable',
    barmode='group',
    title="Ce que je pense de moi vs Ce que les autres pensent de moi",
    labels={'value': 'Note moyenne (sur 10)', 'variable': 'Type d’évaluation'},
    color_discrete_map={'Moyenne_Moi': MALE, 'Moyenne_Autres': FEMALE},
    width=1400,
    height=600
)

fig.update_layout(xaxis_tickangle=-45)
fig.show()
```
