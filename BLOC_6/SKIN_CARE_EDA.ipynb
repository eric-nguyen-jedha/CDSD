```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os
import pandas as pd

# Paramètres
IMAGES_PATH_1 = "/Users/eric/Desktop/JEDHA/CERTIF/09_SKIN_CARE/archive/HAM10000_images_part_1"
IMAGES_PATH_2 = "/Users/eric/Desktop/JEDHA/CERTIF/09_SKIN_CARE/archive/HAM10000_images_part_2"

# Fonction pour obtenir le chemin correct (recherche dans les deux dossiers)
def get_image_path(image_id):
    # Essayer dans part_1
    path1 = os.path.join(IMAGES_PATH_1, f"{image_id}.jpg")
    if os.path.exists(path1):
        return path1
    
    # Essayer dans part_2
    path2 = os.path.join(IMAGES_PATH_2, f"{image_id}.jpg")
    if os.path.exists(path2):
        return path2
    
    # Si aucune ne fonctionne
    print(f"❌ Image non trouvée : {image_id}")
    return None

# Définition des classes
CLASS_LABELS = {
    0: 'akiec',
    1: 'bcc',
    2: 'bkl',
    3: 'df',
    4: 'nv',
    5: 'vasc',
    6: 'mel'
}

# Créer un dictionnaire {class_name: path}
class_to_image = {}

print("🔍 Vérification des images par classe :\n")
for class_name in CLASS_LABELS.values():
    subset = df[df['dx'] == class_name]
    if not subset.empty:
        image_id = subset.iloc[0]['image_id']
        img_path = get_image_path(image_id)
        
        if img_path is not None:
            class_to_image[class_name] = img_path
            print(f"✅ {class_name}: {image_id}.jpg → trouvé")
        else:
            print(f"❌ {class_name}: {image_id}.jpg → NON trouvé")
            class_to_image[class_name] = None
    else:
        print(f"❌ {class_name}: Aucune image dans le DataFrame")
        class_to_image[class_name] = None

# Afficher les images
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.ravel()

for i, (class_name, img_path) in enumerate(class_to_image.items()):
    if i < len(axes):
        if img_path is not None:
            try:
                img = mpimg.imread(img_path)
                axes[i].imshow(img)
                axes[i].set_title(class_name.upper(), fontsize=10)
                axes[i].axis('off')
            except Exception as e:
                axes[i].text(0.5, 0.5, f"Erreur", ha='center', va='center', color='red')
                axes[i].axis('off')
                print(f"Erreur lors du chargement de {img_path}: {e}")
        else:
            axes[i].text(0.5, 0.5, f"{class_name.upper()}\nImage non trouvée", ha='center', va='center', color='red', fontsize=10)
            axes[i].axis('off')

# Masquer les axes restants
for i in range(len(class_to_image), len(axes)):
    fig.delaxes(axes[i])

plt.suptitle("Exemples d'images par classe", fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```

```python
import plotly.express as px
import pandas as pd

# 1. Nettoyage global
df_clean = df.dropna(subset=['localization']).copy()

# 2. Distribution générale (toutes lésions)
total_counts = df_clean['localization'].value_counts().reset_index()
total_counts.columns = ['Localization', 'Total']
total_counts['Total_Perc'] = (total_counts['Total'] / total_counts['Total'].sum()) * 100

# 3. Distribution mélanome
melanoma_df = df_clean[df_clean['dx'] == 'mel']
mel_counts = melanoma_df['localization'].value_counts().reset_index()
mel_counts.columns = ['Localization', 'Mel_Count']
mel_counts['Mel_Perc'] = (mel_counts['Mel_Count'] / len(melanoma_df)) * 100

# 4. Fusionner
merged = pd.merge(total_counts, mel_counts, on='Localization', how='left').fillna(0)

# 5. Trier par ordre décroissant du total
merged = merged.sort_values('Total', ascending=False)

# 6. Préparer pour Plotly : données longues
df_plot = pd.melt(
    merged,
    id_vars=['Localization'],
    value_vars=['Total_Perc', 'Mel_Perc'],
    var_name='Type',
    value_name='Percentage'
)

# 7. Légende claire
df_plot['Type'] = df_plot['Type'].map({
    'Total_Perc': 'Toutes lésions',
    'Mel_Perc': 'Mélanome'
})

# 8. Palette : utiliser une couleur de Plasma pour le mélanome
# On choisit un rouge/orange vif de la palette Plasma
plasma_red = px.colors.sequential.Plasma[7]  # Orange-rouge vif

fig = px.bar(
    df_plot,
    x='Localization',
    y='Percentage',
    color='Type',
    barmode='group',
    title='Comparaison : Distribution des Localisations<br><sup>Ensemble du dataset vs Mélanomes – En pourcentage</sup>',
    labels={'Percentage': 'Pourcentage (%)', 'Localization': 'Localisation'},
    color_discrete_map={
        'Toutes lésions': 'lightgray',
        'Mélanome': plasma_red  # 🔥 Couleur vive de la palette Plasma
    },
    text='Percentage'
)

# 9. Formater le texte
fig.update_traces(
    texttemplate='%{y:.1f}%',
    textposition='outside',
    textfont_size=10
)

# 10. Mise en page
fig.update_layout(
    xaxis_title="Localisation anatomique",
    yaxis_title="Pourcentage (%)",
    template="plotly_white",
    width=1000,
    height=550,
    xaxis=dict(tickangle=45),
    yaxis=dict(range=[0, max(df_plot['Percentage']) * 1.15]),
    legend_title_text="Catégorie",
    title_x=0.5,
    uniformtext_minsize=8,
    uniformtext_mode='hide'
)

# 11. Afficher
fig.show()
```
